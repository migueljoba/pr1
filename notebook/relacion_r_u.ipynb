```python
# importar librerías para la simulación 
import numpy as np
import pandas as pd
import utils
import utils_plotly
from rule import RulePgg
```

#### Programa

```python
def run(rule: RulePgg, generations=50, sides=101):
    initial_population = utils.random_population([0, 1], [0.5, 0.5], (sides, sides))

    matrix_list = utils.run_pgg(initial_population, rule, generations, stop_when_all=0)

    freq_data = utils.resume_frequency_data(matrix_list, strategy=[1])

    # plot_title = f"aporte: {rule.pay}, tolerancia: {rule.tolerance}, r: {rule.factor}"
    # plot = utils_plotly.plot_frequency(data=freq_data, title=plot_title)
    # plot.show()

    return freq_data


rule = RulePgg()
rule.pay = 1
sides = 11
frequency_table_data = []
factor_table_data = []
factor_range = utils.custom_range(3, 5, step=0.1)
tolerance_range = utils.custom_range(0, 50, step=1)

for t in tolerance_range:
    rule.tolerance = t
    row_tolerance = []

    for r in factor_range:
        print(".", end="")
        rule.factor = r
        freq_data = run(rule, sides=sides)

        row_tolerance.append(len(freq_data))

    factor_table_data.append(row_tolerance)
```

## Relación entre factor multiplicador, tolerancia y equilibrio

Cada celda indica en cuántas generaciones se alcanza la invasión total para valores de tolerancias (%) *u* y factores multiplicadores *r* dados.  
El objetivo es encontrar el número más alto de celda, con los menores valores posibles de *u* y *r*.

En la tabla, cada fila representa corridas del programa con tolerancia fija (u) y factor multiplicador (r) variable.  
 
Las celdas con valor >100 indican que en 100 generaciones no se alcanzó invasión total o se alcanzó el equilibrio, se detuvo la simulación y se avanzó a otro valor de *r* o *u*.  


#### (OBS.: Se debe comprobar afirmaciones con más simulaciones y distinas poblaciones iniciales)
##### Se verifica que:
Para cualquier valor de tolerancia u, la invasión total se retrasa mientras mayor sea el factor multiplicador r.

 Para cualquier valor de tolerancia u, la invasión total se retrasa mientras mayor sea el factor multiplicador r. 






```python
def gen_condition(value):
    return ">100" if value >= 100 else value


def apply_style(styler):
    styler.set_caption("r: factor multiplicador u: tolerancia (%)")
    styler.format(gen_condition)
    styler.background_gradient(axis=None, vmin=1, vmax=100, cmap="Reds")
    return styler


pd.set_option('display.max_rows', None)

dataframe = pd.DataFrame(factor_table_data,
                         columns=[f"r = {r}" for r in factor_range],
                         index=[f"u = {t}" for t in tolerance_range])

# imprimir tabla
dataframe  #.style.pipe(apply_style)
```

```python
import numpy as np
import pandas as pd
import utils
import utils_plotly
from rule import RulePgg


def run(rule: RulePgg, generations=100, sides=20):
    initial_population = utils.random_population([0, 1], [0.2, 0.8], (sides, sides))

    matrix_list = utils.run_pgg(initial_population, rule, generations, stop_when_all=0)

    freq_data = utils.resume_frequency_data(matrix_list, strategy=[1])

    # plot_title = f"aporte: {rule.pay}, tolerancia: {rule.tolerance}, r: {rule.factor}"
    # plot = utils_plotly.plot_frequency(data=freq_data, title=plot_title)
    # plot.show()

    # mapa de evolucion, animado
    # utils_plotly.imshow_animate(np.array(matrix_list))
    return freq_data


rule = RulePgg()
rule.pay = 1

frequency_table_data = []
factor_table_data = []
factor_range = utils.custom_range(1, 5, step=0.5)
tolerance_range = utils.custom_range(0, 50, step=2)

for r in factor_range:

    print(f"r = {r}" + "#" * 20)

    rule.factor = r
    factor_table_row = []

    for t in tolerance_range:
        rule.tolerance = t
        freq_data = run(rule, sides=21)
        frequency_table_data.append(freq_data)

        factor_table_row.append(len(freq_data))

    factor_table_data.append(factor_table_row)

pd.set_option('display.max_rows', None)

pd.DataFrame(factor_table_data,
             index=[f"r = {r}" for r in factor_range],
             columns=[f"t = {t}" for t in tolerance_range])
```

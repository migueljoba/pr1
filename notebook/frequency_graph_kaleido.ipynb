### Invasión de desertores - Gráfico de frecuencia

#### Objetivo
Tomando un valor de incentivo **b** arbitrario, se desea conocer cuántas generaciones son necesarias 
para que la estrategia cooperadora se extinga.  
Se corre el algoritmo con diferentes tamaños de población.


#### Condiciones de las evoluciones
- Incentivo **b** arbitrario igual a 1.85.
- Población cuadrada, representada por matriz de orden NxN, para N impar.
- Inicialmente existe un único desertor, ubicado exactamente en el centro de la matriz. El resto de la población es cooperadora.
- Cada corrida tendrá 350 generaciones.
- N aumenta en 2 para cada corrida, siendo:  
$
  N_{0}=3, N_{1}=5, N_{2}=7, ..., N_{n-1}=47, N_{n}=49;
$


```python
import utils
import time
import utils_plotly
import utils_population
from rule import Rule
```

```python

sides_values = range(3, 50, 2)  # (inicio, fin, paso) -> lista de N impares, desde 3 a 49

for sides in sides_values:
    
    # definir reglas de evolucion
    rule = Rule()
    rule.b = 1.85
    rule.use_binary_rule()
    rule.use_binary_transition()
    generations = 350

    initial_population = utils_population.single_defector(sides=sides)
    matrix_list = utils.run(initial_population, rule, generations)

    plot_data = utils.resume_frequency_data(matrix_list)

    plot_title = f"b: {rule.b}, dim: {initial_population.shape}"
    plot = utils_plotly.plot_frequency(data=plot_data, title=plot_title)
    time.sleep(0.1)
    plot.show()
```

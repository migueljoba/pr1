## Equilibrio en caleidoscopio
Relacionado a Invasión de desertores

```python
import numpy as np

import utils
import utils_plotly
import utils_population
from rule import Rule

# rango para b [1, 2]
[1.15, 1.2, 1.25, 1.3, 1.35, 1.5, 1.55, ]

[1.4, 1.45]


def run(b: float, sides: int = 7, generations: int = 20):
    rule = Rule()
    rule.b = b
    rule.use_4s_rule()
    rule.use_4s_transition()
    initial_population = utils_population.single_defector(sides=sides)
    matrix_list = utils.run(initial_population, rule, generations)
    
    # grafico de lineas
    plot_data = utils.resume_frequency_data(matrix_list)
    plot_title = f"b: {rule.b}, dim: {initial_population.shape}, gen.: {generations}"
    plot = utils_plotly.plot_frequency(data=plot_data, title=plot_title)
    plot.show()
    
    # mapa de evolucion, animado
    utils_plotly.imshow_animate(np.array(matrix_list))
```

#### b = 1.5
